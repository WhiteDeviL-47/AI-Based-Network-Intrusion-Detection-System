```python
### 1. Import Required Libraries

import numpy as np

from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

import joblib

print("Libraries imported successfully")
```

```python
### 2. Load Preprocessed Training and Testing Data

X_train = np.load("X_train.npy")
X_test  = np.load("X_test.npy")
y_train = np.load("y_train.npy")
y_test  = np.load("y_test.npy")

print("Training data shape:", X_train.shape)
print("Testing data shape :", X_test.shape)
```

```python
### 3. Train Random Forest Classifier

from sklearn.ensemble import RandomForestClassifier
import joblib

rf_model = RandomForestClassifier(
    n_estimators=200,
    class_weight={0: 1, 1: 3},  # prioritize attacks
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

joblib.dump(rf_model, "random_forest_model_tuned.pkl")


print("Random Forest training completed")
```

```python
rf_predictions = rf_model.predict(X_test)
rf_accuracy = accuracy_score(y_test, rf_predictions)

print("Random Forest Accuracy:", rf_accuracy)
```

```python
### 5. Save Random Forest Model

joblib.dump(rf_model, "random_forest_model.pkl")
print("Random Forest model saved")
```

```python
### 6. Train Support Vector Machine (SVM)

svm_model = SVC(kernel='rbf', gamma='scale')

svm_model.fit(X_train, y_train)

print("SVM training completed")
```

```python
### 7. Evaluate SVM Model

svm_predictions = svm_model.predict(X_test)
svm_accuracy = accuracy_score(y_test, svm_predictions)

print("SVM Accuracy:", svm_accuracy)
```

```python
### 8. Save SVM Model

joblib.dump(svm_model, "svm_model.pkl")
print("SVM model saved")
```
