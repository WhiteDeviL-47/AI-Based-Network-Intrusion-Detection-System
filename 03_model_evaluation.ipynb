```python
### 1. Import Required Libraries

import numpy as np
import joblib

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully")
```

```python
### 2. Load Preprocessed Test Data

X_test = np.load("X_test.npy")
y_test = np.load("y_test.npy")

print("Test data loaded")
```

```python
### 3. Load Trained Machine Learning Models

rf_model = joblib.load("random_forest_model_tuned.pkl")
svm_model = joblib.load("svm_model.pkl")

print("Models loaded successfully")
```

```python
### 4. Evaluate Random Forest Model Performance

rf_pred = rf_model.predict(X_test)

rf_accuracy = accuracy_score(y_test, rf_pred)
print("Random Forest Accuracy:", rf_accuracy)

print("\nRandom Forest Classification Report:\n")
print(classification_report(y_test, rf_pred))
```

```python
### 5. Random Forest Confusion Matrix

rf_cm = confusion_matrix(y_test, rf_pred)

plt.figure(figsize=(5, 4))
sns.heatmap(rf_cm, annot=True, fmt="d", cmap="Blues")
plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.show()
```

```python
### 6. Evaluate Support Vector Machine (SVM) Model Performance

svm_pred = svm_model.predict(X_test)

svm_accuracy = accuracy_score(y_test, svm_pred)
print("SVM Accuracy:", svm_accuracy)

print("\nSVM Classification Report:\n")
print(classification_report(y_test, svm_pred))
```

```python
### 7. Support Vector Machine (SVM) Confusion Matrix

svm_cm = confusion_matrix(y_test, svm_pred)

plt.figure(figsize=(5, 4))
sns.heatmap(svm_cm, annot=True, fmt="d", cmap="Greens")
plt.title("SVM Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.show()
```
